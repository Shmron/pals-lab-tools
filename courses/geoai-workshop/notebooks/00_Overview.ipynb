## Why does this course exist?

Climate is changing. In sub-Saharan Africa, that change is not abstract —
it shows up in longer dry seasons, earlier and more intense heat spikes,
and rainfall that arrives later or fails entirely.

What happens inside a pregnancy when a mother lives through those conditions?
Does sustained drought in the first trimester affect fetal growth?
Does heat in the final weeks of pregnancy change birth outcomes?

These are not hypothetical questions. They have answers — but only if you can
**connect the right data to the right people at the right time.**

That is what this course teaches you to do.

## The data: Zimbabwe DHS 2015

Throughout this course we work with one real dataset:
the **Zimbabwe Demographic and Health Survey (DHS), 2015**.

The DHS is a nationally representative household survey conducted across
more than 90 countries. It collects detailed information on fertility,
maternal and child health, nutrition, and household characteristics.

The Zimbabwe 2015 survey covered women aged 15–49 and recorded information
on all births in the five years preceding the survey.
We will work with the birth records from that survey.

Every birth record tells us:
- When the baby was born (and when conception likely occurred)
- Where the mother's household was located (at cluster level)
- Whether the baby was born with low birth weight, preterm, or small for gestational age
- The mother's age, education level, and household wealth
- Whether she attended antenatal care and where she delivered

What the DHS does **not** tell us is what the weather was like during the pregnancy.
That is what the satellites are for.


## The environmental data: from satellites to numbers

Zimbabwe has a relatively sparse network of ground weather stations.
For research that needs to cover the whole country over multiple years,
we rely on **Earth observation** products:

| Source | What it measures | Spatial resolution |
|--------|-----------------|-------------------|
| **ERA5** (ECMWF reanalysis) | Air temperature, heat stress | ~25 km |
| **CHIRPS** | Rainfall | ~5 km |
| **MODIS** | Vegetation index (NDVI) | 500 m–1 km |
| **SPEI** | Drought index | ~50 km |

These products combine satellite observations with ground data and
atmospheric models to produce gridded, consistent time series going back decades.

They are not perfect. They have spatial and temporal resolution limits.
They are modelled estimates, not direct measurements.
Understanding those limitations is part of the course.

## What is GeoAI?

**GeoAI** is the use of artificial intelligence — particularly large language models —
to work with geospatial and health data in more natural and accessible ways.

In this course, GeoAI appears in the final notebook as a practical tool:
a natural-language interface that lets you ask questions about the dataset in plain English
and receive both a verified SQL query and an immediate result.

Rather than memorising database query syntax, you describe what you want to know.
The AI drafts the query, shows it to you for review, and then runs it.

This is not a replacement for understanding the data — it is an accelerator for exploration.
You still need to judge whether the query makes sense and whether the result is plausible.
That judgement comes from everything you will learn in notebooks 1 through 5.
